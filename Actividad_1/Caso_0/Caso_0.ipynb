### Asignación de variables
La magia de la programación y su capacidad de automatización se potencian cuando le asignamos una cantidad a una variable definida por nosotros. Este proceso se conoce como *asignación* y tiene la siguiente estructura:

*nombre de variable* = *cantidad a asignar*

Por ejemplo, observemos las siguientes asignaciones:

```python
# Declaración de variables básicas con anotaciones de tipo estático (Type Hints)
A: int = 550
B: int = 6
C: str = 'Hola'
D: str = "Amigos"
E: bool = True
F: bool = False
```

Veamos que ocurre cuando llamamos cada una de las variables que hemos definido


```python
print(A)
print(B)
print(C)
print(D)
print(E)
print(F)
```

Veamos qué ocurre cuando llamamos una variable que no hemos definido previamente:

```python
try:
    a
except NameError as e:
    print("[EXPLICACIÓN ACADÉMICA] Ocurrió un NameError esperado:")
    print(f"  Detalle: {e}")
    print("  Explicación: En Python, las variables son sensibles a mayúsculas y minúsculas (case-sensitive).")
    print("  La variable 'A' (mayúscula) está definida, pero 'a' (minúscula) no existe en el namespace actual.")
```

Este es un error muy común en Python y en cualquier lenguaje de programación, siempre que llamemos una variable esta debe haber sido previamente definida.

Las variables son alias para las cantidades que les han sido asignadas y por lo tanto tienen las mismas propiedades y operadores.

Veamos:

```python
A+B
```

```python
C+D
```

```python
C+' '+D
```

```python
print(E & F)   # operación lógica Y
print(E | F)   # operación lógica O
```

```python
E & F
E | F
```

Este ejemplo sencillo nos muestra cómo el efecto del operador `+` depende de la naturaleza de las cantidades que se están operando. 
- En el caso de números, el operador `+` representa una suma numérica. 
- En el caso de *strings* (así llamamos a las variables de texto en Python) el operador `+` representa una concatenación, es decir, crear una nueva *string* que resulta de juntar las dos *strings* que se están concatenando.

Es muy importante tener en cuenta que cada tipo de variable diferente en Python tiene definidos un conjunto de operadores que representan operaciones diferentes en cada caso. Un ejemplo son las variables booleanas con valores `True` o `False` que tienen definidos los operadores `&` y `|` para ejecutar las operaciones lógicas Y y O, respectivamente.

En ciencia de datos es común obtener números en formato de texto, el paso de texto a número se logra de manera muy sencilla utilizando las funciones `int()` o `float()`, dependiendo de si el número es un entero o un flotante.

Hagamos las siguientes definiciones:

```python
# Declaración de variables con tipos mixtos (entero y cadena de texto)
A: int = 3
B: str = '5'
```

Intentemos ahora sumar estos dos números:

```python
try:
    A + B
except TypeError as e:
    print("[EXPLICACIÓN ACADÉMICA] Ocurrió un TypeError esperado:")
    print(f"  Detalle: {e}")
    print("  Explicación: Python es un lenguaje de tipado dinámico pero fuertemente tipado.")
    print("  No permite realizar operaciones implícitas entre tipos incompatibles como enteros (int) y cadenas (str).")
```

Recibimos un error, esto es debido a que Python no tiene definido un operador `+` para operar enteros, `A`, y strings, `B`. Para realizar la suma debemos primero transformar el string `B` en un número:

```python
A+int(B)
```

O alternativamente

```python
# Casting explícito de tipo cadena (str) a tipo entero (int)
C: int = int(B)
resultado_suma: int = A + C
print(f"Resultado de la suma tipada (esperado 8): {resultado_suma}")
```

Como en ningún momento hemos modificado a la variable `B`, este debe ser aún de tipo string.

```python
B
```

Miremos a las diferencias entre estas distintas operaciones:

```python
print('3.1415926')
try:
    print(int('3.1415926') + 2)
except ValueError as e:
    print("[EXPLICACIÓN ACADÉMICA] Ocurrió un ValueError esperado:")
    print(f"  Detalle: {e}")
    print("  Explicación: La función int() no puede convertir directamente strings con formato decimal.")
    print("  Se debe realizar float() primero, o usar float('3.1415926') + 2.")
    print(f"  Solución corregida: {int(float('3.1415926')) + 2}")
print(int('5'))
```

### Funciones
La automatización de procesos en programación está potenciada por la definición de funciones. Una función tiene 4 componentes principales:
- Un nombre que la identifique y que usaremos para llamarla.
- Un conjunto de entradas, que será la información que la función tomará para llevar a cabo sus procesos.
- Un proceso, que es un conjunto de instrucciones dadas para procesar la información de las entradas y obtener el resultado final. Este es el corazón de la función.
- Una salida, que es un conjunto de datos que la función arrojará como el resultado de su proceso de ejecución.

Cuando abrimos un Notebook de Python, hay un conjunto de funciones predefinidas que podemos utilizar, algunas de estas ya las hemos usado antes:
- `print` esta es una función que nos sirve para imprimir en la pantalla los valores que se entregaron como entrada.
- `int`: esta es una función que nos sirve para convertir en entero una string que tenga el formato de un número entero.
- `float`: esta es una función que nos sirve para convertir en número flotante una string que tenga el formato de número.

```python
# Declaración tipada de datos personales
nombre: str = "Miguel Angel Uribe Laverde"
print(f"Nombre: {nombre}")
edad: str = "40"
print(f"Edad: {edad}")
print(f"Clase de la variable 'edad': {type(edad)}")
```

La función `type` es la función a la que debemos recurrir siempre que deseemos conocer a qué clase pertenece una variable determinada.

A estas alturas ya estamos familiarizados con cómo se debe llamar una función:

*nombre_funcion ( listado de entradas )*

veamos ahora el formato general para definir una función:


def *nombre_funcion ( listado de entradas ):*<br>
conjunto de operaciones<br>
return *salida_de_la_funcion*

Esto queda más claro con un ejemplo:


```python
from typing import Union

def funcion_suma(x: Union[int, float], y: Union[int, float]) -> Union[int, float]:
    """
    Suma dos valores numéricos escalares.
    
    Args:
        x (int o float): Primer sumando.
        y (int o float): Segundo sumando.
        
    Returns:
        int o float: El resultado de la suma.
    """
    s: Union[int, float] = x + y
    return s
```

```python
try:
    funcion_suma(3, 'ejemplo')
except TypeError as e:
    print("[EXPLICACIÓN ACADÉMICA] Ocurrió un TypeError esperado en la llamada a la función:")
    print(f"  Detalle: {e}")
    print("  Explicación: La operación interna x + y falla al intentar sumar un entero con un string.")
```

Definamos una función que tome dos strings y que cree una nueva string concatenando las dos primeras letras de las entradas:

```python
def iniciales(x: str, y: str) -> str:
    """
    Concatena las dos primeras letras de dos cadenas de texto.
    
    Args:
        x (str): Primera palabra.
        y (str): Segunda palabra.
        
    Returns:
        str: String con las dos primeras iniciales de x e y.
    """
    texto: str = x[0:2]
    texto = texto + y[0:2]
    return texto
```

```python
iniciales("Miguel", "Uribe")
```

```python
# Declaración de variable de texto para manipulación de strings
texto: str = "Miguel Angel Uribe Laverde"
print(f"Texto registrado: '{texto}' (longitud {len(texto)} caracteres)")
```
