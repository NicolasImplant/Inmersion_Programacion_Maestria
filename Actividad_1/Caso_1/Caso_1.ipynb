### Conjuntos
Para comenzar a resolver nuestro problema, el primer tipo de variable que usaremos será un conjunto (*set* en inglés).

Los conjuntos se pueden ver como colecciones de elementos (que pueden ser de diferente tipo) con las siguientes propiedades:
- No puede haber elementos repetidos
- No hay un orden entre los elementos
- Los elementos en un conjunto no se pueden modificar

Los conjuntos son muy útiles cuando se desea comparar colecciones de elementos.


Definamos cuatro conjuntos diferentes para incluir la información de las características importantes de cada uno de los apartamentos:

```python
from typing import Set

# Definición tipada de las características de cada propiedad
ed_santa_barbara_carac: Set[str] = {'porteria 24 horas', 'balcón', 'piscina', 'gimnasio', 'terraza'}
ed_porvenir_carac: Set[str] = {'porteria 24 horas', 'balcón', 'piscina'}
ed_calasanz_carac: Set[str] = {'porteria 24 horas', 'balcón', 'gimnasio'}
ed_alcala_carac: Set[str] = {'balcón', 'piscina', 'terraza'}
```

Para definir un conjunto se utilizan llaves `{` `}` y los elementos se separan con comas `,`.

¡Ups! Probablemente haya notado que omitimos la caldera en el conjunto de propiedades del apartamento en el Edificio Santa Bárbara. Esto se puede solucionar facilmente con el comando `add`:

```python
ed_santa_barbara_carac.add('caldera')
ed_santa_barbara_carac
```

#### Intersección de Conjuntos
Para calcular la intersección de conjuntos utilizamos el operador `&`.

```python
ed_santa_barbara_carac & ed_porvenir_carac & ed_calasanz_carac & ed_alcala_carac
```

#### Unión entre conjuntos
Para calcular la unión entre conjuntos utilizamos el operador `|`.

![Picture title](fig/image-20211025-125443.png)


Veamos el conjunto total de características observadas en los apartamentos de interés:

```python
ed_santa_barbara_carac | ed_porvenir_carac | ed_calasanz_carac | ed_alcala_carac
```

Nótese que no hay características repetidas, ya que no es posible tener elementos repetidos en los conjuntos.

#### Diferencia de conjuntos
La diferencia de conjuntos `A-B` es el resultado de tomar los elementos del conjunto `A` y remover de ese grupo los elementos que se encuentren también en el elemento `B`. Para hacer esta operación utilizamos el operador `-`. 

![Picture title](fig/image-20211025-143803.png)

Usemos la diferencia de conjuntos para encontrar las características del apartamento en el Edifio Calasanz que no tiene el apartamento del Edificio Porvenir:

```python
ed_calasanz_carac - ed_porvenir_carac
```

La única característica presente en el Edificio Calasanz que no está en el Edificio Porvenir es la piscina.

**Ojo:** Esta operación no es conmutativa, si invertimos los argumentos obtendremos las propíedades del apartamento en el Edificio Porvenir, que no tenemos en el Edificio Calasanz.

```python
ed_porvenir_carac - ed_calasanz_carac
```

Esta misma operación la podemos usar para calcular, por ejemplo, cuáles de las propiedades del apartamento en el Edificio Santa Bárbara son exclusivas de esta propiedad:

```python
ed_santa_barbara_carac - ed_porvenir_carac - ed_calasanz_carac - ed_alcala_carac
```

El apartamento en Edificio Santa Bárbara es el único entre las opciones de interés que cuenta con caldera.

Los conjuntos también facilitan mucho la evaluación de la pertenencia o no de un elemento. Por ejemplo, si quisiéramos saber si el apartamento del Edificio Porvenir tiene piscina:

```python
'piscina' in ed_porvenir_carac
```

El resultado es un booleano que nos indica si es cierto (`True`) o falso (`False`) que el elemento pertenece a un conjunto.

### Listas

Las listas son otro tipo de colección de elementos. Al igual que los conjuntos, las listas también pueden contener elementos de diferentes tipos. Sin embargo, las propiedades de las listas son diferentes:

- Sí hay un ordenamiento de elementos y sí es posible acceder a los elementos individualmente.
- Sí es posible modificar los elementos en una lista.
- Sí es posible repetir elementos en una lista.

Usemos una lista para incluir la información de los teléfonos de contacto de la propiedad del Edificio Santa Bárbara:

```python
from typing import List

# Lista de teléfonos de contacto del Edificio Santa Bárbara
ed_santa_barbara_tels: List[str] = ['325 2589879', '326 3547845', '325 4598985']
```

Note que para definir la lista hemos usado corchetes `[` `]` y que los elementos de la lista se separan con comas `,`. Los números de teléfono han sido ingresados como strings ya que no nos interesa hacer operaciones numéricas con ellos en el futuro.<br>
Imprimamos ahora la lista:

```python
ed_santa_barbara_tels
```

En las listas son colecciones ordenadas, de manera que el orden en el que incluimos los elementos es el mismo en el que se presentan al imprimir la lista. Es posible, además acceder a cada uno de los elementos usando un índice que arranca en 0 y termina en `N-1`, en donde `N` es el total de elementos de la lista.

Revisemos las siguientes salidas:

```python
print("Elemento 0 de la lista:", ed_santa_barbara_tels[0])
print("Elemento 1 de la lista:", ed_santa_barbara_tels[1])
print("Elemento 2 de la lista:", ed_santa_barbara_tels[2])
try:
    print("Elemento 3 de la lista:", ed_santa_barbara_tels[3])
except IndexError as e:
    print("\n[EXPLICACIÓN ACADÉMICA] Ocurrió un IndexError esperado:")
    print(f"  Detalle: {e}")
    print("  Explicación: En Python las listas inician en el índice 0.")
    print(f"  Con {len(ed_santa_barbara_tels)} elementos, los índices válidos van de 0 a {len(ed_santa_barbara_tels)-1}.")
    print("  Intentar acceder al índice 3 está fuera de los límites de la colección.")
```

La última línea arroja un error debido a que no existe el elemento 3 en la lista.

**Importante** Para llamar elementos de una colección, independiente del tipo de colección, el índice se ubica siempre entre corchetes `[` `]`.

Alternativamente, también podemos usar índices negativos para recorrer la lista de atrás hacia adelante. Así, el elemento -1 siempre corresponderá al último elemento de la lista, el -2 corresponderá al penúltimo elemento, y así sucesivamente.

```python
print("Elemento -1 de la lista:", ed_santa_barbara_tels[-1])
print("Elemento -2 de la lista:", ed_santa_barbara_tels[-2])
print("Elemento -3 de la lista:", ed_santa_barbara_tels[-3])
```

¡Ups! Posiblemente te diste cuenta de que cometimos un error al ingresar el tercer teléfono de contacto, el valor que aparece en nuestro cuaderno de anotaciones es 325 4589895, mientras que en el listado de teléfonos tenemos:

```python
ed_santa_barbara_tels[-1]
```

La buena noticia es que las listas si son mutables, y sí podemos modificar sus elementos. Para hacer la actualización simplemente redefinimos el último elemento de la lista de la siguiente manera:

```python
ed_santa_barbara_tels[-1] = '325 4589895'
ed_santa_barbara_tels
```

Creemos ahora una lista completa con toda la información que tenemos de la propiedad en el Edificio Santa Bárbara:

```python
from typing import Any, List

# Lista que agrupa información mixta de la propiedad
ed_santa_barbara_list: List[Any] = [
    'Calle 127 # 7 - 15, AP208',  # Dirección
    ed_santa_barbara_tels,       # Sublista de teléfonos
    3,                           # Habitaciones
    2,                           # Baños
    2,                           # Garajes
    3750000,                     # Valor arriendo
    85,                          # Área m²
    ed_santa_barbara_carac       # Características (conjunto)
]
```

Los elementos de la lista que acabamos de crear son de muchas características diferentes. Por ejemplo, el primer elemento corresponde a la dirección y está guardada en forma de string, el segunda elemento es una lista con los teléfonos de contacto, los siguientes elementos tienen formato numérico y el último elemento es un conjunto con las propiedades del apartamento.

Veamos cómo acceder a diferentes elementos de la información guardada en la lista que acabamos de crear:

```python
print("El número de habitaciones: ",ed_santa_barbara_list[2])
print("El número de garajes: ",ed_santa_barbara_list[4])
print("El valor del arrendamiento: ", ed_santa_barbara_list[5])
print("El área del apartamento: ",ed_santa_barbara_list[-2])
print("El tercer número de contacto ", ed_santa_barbara_list[1][2])
```

En Python es muy natural la forma de llamar elementos en listas que están dentro de listas: `ed_santa_barbara_list[1]` accede al segundo elemento de la lista `ed_santa_barbara_list`. Como este elemento es una lista, se puede acceder al tercer elemento de la misma usando un nuevo índice `ed_santa_barbara_list[1][2]`.

Supongamos ahora que para comparar apartamentos, es importante también incluir la información del número de piso. Debemos entonces agregar un elemento adicional a nuestra lista con esta información, esto se logra con la función `append()`:

```python
ed_santa_barbara_list.append(2)  # el número de piso es 2
ed_santa_barbara_list   # imprimimos de nuevo para ver el resultado
```

Otra operación posible con las listas es la *concatenación*, esto es, la unión sucesiva de dos listas para formar una lista más grande. Esta operación se puede lograr con el operador `+`. Veamos:

```python
ed_santa_barbara_list + ['otra_info', 4545, {'random', 'info', 25}]
```

### Conjuntos y Listas
Es posible crear conjuntos a partir de listas, y viceversa. Para crear conjuntos a partir de una lista (o de cualquier otra colección) se usa la función `set()`, esta es una aproximación muy útil cuando se desea conocer el número de elementos diferentes en una lista.

```python
A = [1,2,3,2,3,1,2,3,3,2,1,2,3]  # creamos una lista de elementos
Aset = set(A)                    # creamos un conjunto a partir de la lista
print(Aset)                      # El conjunto sólo tiene los elementos diferentes
print("El número de elementos diferentes en la lista: ", len(Aset))
```

También es posible crear listas a partir de conjuntos (o cualquier otra colección), para esto se usa la función `list()`.

```python
B = {2, 4, 6, 8, 10, 12}          # creamos un conjunto
Blist = list(B)                   # creamos una lista a partir del conjunto
Blist.append(6)                   # agregamos un nuevo elemento a la lista
print(Blist)
```

### Diccionarios

Un diccionario es una colección de elementos que, como en conjuntos y listas, pueden ser de diferente tipo. Su característica principal es que los índices no son numéricos, sino que pueden corresponder a otros elementos como strings. Otras propiedades de los diccionarios son:

- No tienen ordenamiento debido a que los índices no son numéricos.
- Sí pueden tener elementos repetidos.
- Sí es posible modificar sus elementos.

Veamos como incluir la información del apartamento del Edificio Santa Bárbara en un diccionario:

```python
from typing import Any, Dict

# Diccionario indexado por características del apartamento
ed_santa_barbara_dict: Dict[str, Any] = {
    'direccion': 'Calle 127 # 7 - 15, AP208',
    'telefonos': ed_santa_barbara_tels,
    'n_habitaciones': 3,
    'n_piso': 2,
    'n_garajes': 2,
    'valor_arrendamiento': 3750000,
    'area': 85,
    'caracteristicas': ed_santa_barbara_carac
}
```

Para definir un diccionario de nuevo usamos llaves `{` `}` y los elementos de nuevo están separados por comas `,`. Sin embargo, en un diccionario cada elemento está definido por una *llave* (*key* en inglés) y un *item* separados por un `:`.

Veamos el resultado de la definición anterior.

```python
ed_santa_barbara_dict
```

Ahora, para acceder a diferentes características usamos la *llave* correspondiente:

```python
print('El área del apartamento: ', ed_santa_barbara_dict['area'])
print('El número de garajes: ', ed_santa_barbara_dict['n_garajes'])
print('Las características del apartamento: ', ed_santa_barbara_dict['caracteristicas'])
```

Note los corchetes `[` `]` utilizados para usar las llaves en cada caso.


De nuevo, agreguemos un nuevo parámetro correspondiente al número de piso, en un diccionario esto se logra de la siguiente manera:

```python
ed_santa_barbara_dict['n_piso']=2
ed_santa_barbara_dict
```

Creemos ahora un diccionario con toda la información disponible de los apartamentos de interés:

```python
from typing import Any, Dict

# Estructura de datos anidada que agrupa todas las propiedades inmobiliarias
apartamentos: Dict[str, Dict[str, Any]] = {
    'santa_barbara': {
        'direccion': 'Calle 127 # 7 - 15, AP208',
        'telefonos': ed_santa_barbara_tels,
        'n_habitaciones': 3,
        'n_piso': 2,
        'n_garajes': 2,
        'valor_arrendamiento': 3750000,
        'area': 85,
        'caracteristicas': ed_santa_barbara_carac
    },
    'porvenir': {
        'direccion': 'Calle 135 # 15 - 24, AP703',
        'telefonos': ['324 2785484', '324 3548987'],
        'n_habitaciones': 2,
        'n_piso': 7,
        'n_garajes': 1,
        'valor_arrendamiento': 2700000,
        'area': 75,
        'caracteristicas': ed_porvenir_carac
    },
    'calasanz': {
        'direccion': 'Calle 50 # 80 - 45, AP1201',
        'telefonos': ['324 2548785'],
        'n_habitaciones': 3,
        'n_piso': 12,
        'n_garajes': 2,
        'valor_arrendamiento': 3750000,
        'area': 112,
        'caracteristicas': ed_calasanz_carac
    },
    'alcalá': {
        'direccion': 'Calle 134 # 45 - 15, AP401',
        'telefonos': [],
        'n_habitaciones': 2,
        'n_piso': 4,
        'n_garajes': 2,
        'valor_arrendamiento': 2950000,
        'area': 85,
        'caracteristicas': ed_alcala_carac
    }
}
```

Hemos creado entonces el diccionario `apartamentos`, que tiene cuatro elementos llamados `santa_barbara`, `porvenir`, `calasanz` y `alcalá`. Cada uno de estos elementos es a su vez un diccionario que contiene la información particular de cada propiedad.

Veamos:

```python
print("El área del apartamento en Edificio Calasanz: ", apartamentos['calasanz']['area'])
print("El primer teléfono de contacto del apartamento en Edificio Porvenir: ", apartamentos['porvenir']['telefonos'][0])
print("El número de piso del apartamento en el Edificio Alcalá: ", apartamentos['alcalá']['n_piso'])
```

##### Ejercicio 4
Calcule el valor por metro cuadrado de cada uno de los apartamentos y agregue este valor al listado de parámetros de cada apartamento en el diccionario `apartamento`. ¿Cuál es la propiedad con el menor valor del metro cuadrado?

```python
from typing import Dict, Any

# [RESOLUCIÓN IN-LINE] Bucle para calcular valor_metro e identificar el mínimo
min_val: float = float('inf')
propiedad_min: str = ""

for name, data in apartamentos.items():
    area: float = float(data['area'])
    arriendo: float = float(data['valor_arrendamiento'])
    val_metro: float = arriendo / area
    data['valor_metro'] = val_metro
    print(f"  * {name.title()}: ${val_metro:.2f} por m²")
    if val_metro < min_val:
        min_val = val_metro
        propiedad_min = name

print(f"Propiedad con menor valor m²: '{propiedad_min.title()}' (${min_val:.2f}/m²)")
```

### Tuplas

El último tipo de colección en Python es la tupla (*tuple* en inglés). Este tipo de colección es muy similar a las listas. Sin embargo, difiere de estas en el sentido en que una tupla, o sus elementos, no pueden ser modificados una vez creados. Las propiedades de las tuplas se pueden resumir así:
- Son colecciones ordenadas en las que los elementos se indexan con números.
- No es posible modificar los elementos de una tupla ni tampoco añadir o quitar elementos.
- Las tuplas sí pueden tener elementos repetidos.

Las tuplas entonces se usan en aplicaciones en las que se trabaje con colecciones de objetos que tengan valores fijos y que no se necesiten modificar en el futuro.

Veamos, un ejemplo de creación de una tupla:

```python
ex_tupla = (3, 'arriba', -5, [1,2,3])
```

Los elementos de una tupla se acceden igual que en una lista:

```python
print("primer elemento de la tupla: ", ex_tupla[0])
print("segundo elemento de la tupla: ", ex_tupla[1])
print("tercer elemento de la tupla: ", ex_tupla[2])
print("último elemento de la tupla: ", ex_tupla[-1])
```

Sin embargo, si intentáramos modificar alguno de los elementos de la tupla obtendríamos:

```python
try:
    ex_tupla[0] = 2
except TypeError as e:
    print("[EXPLICACIÓN ACADÉMICA] Ocurrió un TypeError esperado al intentar modificar la tupla:")
    print(f"  Detalle: {e}")
    print("  Explicación: Las tuplas son estructuras inmutables (de solo lectura).")
    print("  No es posible reasignar sus elementos una vez creada la tupla en memoria.")
```

O si intentáramos insertar elementos adicionales:

```python
try:
    ex_tupla.append(8)
except AttributeError as e:
    print("[EXPLICACIÓN ACADÉMICA] Ocurrió un AttributeError esperado:")
    print(f"  Detalle: {e}")
    print("  Explicación: Al ser inmutables, las tuplas no cuentan con métodos modificadores como append().")
```

## Conclusiones

Hemos aprendido sobre las estructuras básicas de Python y sus propiedades:

* **Conjuntos**. Si el orden no importa, se pueden guardar los datos en conjuntos. En conjuntos no se pueden tener elementos repetidos y no es posible acceder a elementos individuales. Por esta misma razón no es posible modificar los elementos existentes en un conjunto. Sí es posible agregar nuevos elementos. Entre conjuntos se pueden hacer operaciones como uniones, intersecciones y diferencias.
* **Listas**. Las listas sí están ordenadas, permiten acceder y modificar los elementos existentes usando índices numéricos. Se pueden eliminar, agregar y modificar los elementos de una lista sin necesidad de crear la lista de nuevo.
* **Tuplas**. Estas son básicamente como listas, con la diferencia de que no pueden ser modificadas. Es decir, no se pueden modificar los elementos existentes en una tupla, y tampoco se pueden quitar o agregar nuevos elementos.
* **Diccionarios**. Los diccionarios también son muy similares a las listas en el sentido de que sus elementos se pueden acceder y modificar, y que es posible agregar y quitar elementos. Sin embargo, los índices para acceder a los elementos son strings y se llaman llaves. No es posible entonces en un diccionario tener llaves repetidas. En un diccionario, los elementos no están ordenados.

- Crear un conjunto a partir de una lista: `conjunto = set(lista)`
- Agregar elemento a conjunto: `conjunto.add(item)`
- Quitar elemento de conjunto: `conjunto.remove(item)`
- Vaciar un conjunto: `conjunto.clear()`
- Obtener el número de elementos del conjunto: `len(conjunto)`
- Unión: `A | B`
- Intersección: `A & B`
- Diferencia: `A-B`
- Diferencia Simétrica: `A ^ B`
- Probar pertenencia: `item in conjunto`

#### Listas
- Definir una lista: `lista = {A, B, C}`
- Crear una lista vacía: `lista = []`
- Crear una lista a partir de una colección: `lista = list(coleccion)`
- Agregar elemento al final lista: `lista.append(item)`
- Quitar elemento de lista por valor: `lista.remove(item)` esta función elimina el primer elemento de la lista igual a `item`.
- Quitar elemento de lista por índice: `del lista[indice]` 
- Obtener la posición del primer elemento con un valor dado: `lista.index(item)`
- Contar el número de veces que un elemento aparece en una lista: `lista.count(value)`
- Obtener un rango de elementos de la lista: `lista[indice_inicial:indice_final]`
- Concatenar listas: `A + B`
- Repetir los contenidos de una lista: `A * N`
- Ordenar los elementos de la lista: `list.sort()`
- Crear una copia de la lista: `B = list(A)`
- Obtener el número de elementos de la lista: `len(lista)`
- Probar pertenencia: `item in lista` 

#### Diccionarios
- Definir un diccionario: `dicc = {'llave1': elemento1, 'llave2': elemento2}`
- Acceder al elemento correpondiente a la llave `key`: `dicc['key']`
- Reemplazar o crear un elemento: `dicc['key'] = item`
- Remover un elemento del diccionario por llave: `del dicc['key']`
- Obtener un listado de todas las llaves en el diccionario: `list(dicc)` o `dicc.keys()`
- Obtener un listado de todos los items del diccionario: `dicc.values()`
- Crear una copia de un diccionario: `B = dicc.copy()`
- Obtener el numero de elementos en el diccionario: `len(dicc)`
- Probar pertenencia: `key in dict` 


#### Tuplas
- Definir una tupla: `tupla = (A, B, C, D)` 
- Definir una tupla conun solo elemento, ojo a la coma al final: `tupla = (A,)`
- Acceder al elemento correspondiente a un índice: `tupla[indice]`
- Obtener el número de elementos de la tupla: `len(tupla)`
- Probar pertenencia: `item in tupla`
