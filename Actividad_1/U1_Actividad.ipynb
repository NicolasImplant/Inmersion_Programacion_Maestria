### Análisis Matemático y de Complejidad:

#### 1. Eliminación de Duplicados y Ordenación (Ejercicio 1.3):
Dada una lista con elementos repetidos, el objetivo es transformarla en una colección de elementos únicos ordenada alfabéticamente.
* **Eliminación de duplicados**: La conversión a `set` se realiza mediante una tabla hash, lo cual tiene una complejidad promedio de $O(N)$ en tiempo.
* **Ordenamiento**: El método `sorted()` implementa el algoritmo **Timsort** (híbrido de Merge Sort e Insertion Sort), con una complejidad de tiempo de $O(N \log N)$ en el peor caso, siendo altamente eficiente y estable.

#### 2. Álgebra de Conjuntos (Ejercicio 1.4):
Basado en el análisis cuantitativo del Diagrama de Venn provisto en `fig/C1_1.png`:

$$
A = \{\text{'a', 'b', 'c', 'e', 'g', 'h', 'l', 'm', 'n', 'o', 'r', 's', 't', 'y'}\}
$$
$$
B = \{\text{'a', 'b', 'c', 'd', 'g', 'h', 'i', 'j', 'm', 'r', 'u', 'w', 'z'}\}
$$

Se computan formalmente las operaciones lógicas de conjuntos:
1. **Intersección ($A \cap B$)**: Elementos que pertenecen simultáneamente a ambos conjuntos (zona central).
   $$A \cap B = \{\text{'a', 'b', 'c', 'g', 'h', 'm', 'r'}\} \quad (\text{7 elementos})$$
2. **Unión ($A \cup B$)**: Total de letras diferentes en ambos conjuntos.
   $$|A \cup B| = 20 \quad (\text{elementos únicos})$$
3. **Diferencia relativa ($B \setminus A$)**: Elementos que pertenecen a $B$ pero no a $A$.
   $$B \setminus A = \{\text{'d', 'i', 'j', 'u', 'w', 'z'}\} \quad (\text{6 elementos})$$

Sin embargo, a diferencia del caso, queremos hacer un análisis un poco más específico en el que, dependiendo de la diferencia del precio de la gasolina en la estación con el promedio del Departamento, a cada estación se le agrege uno de los siguientes códigos:

![Picture title](fig/C2_2.png)

En la figura $\bar{p}$ representa el valor promedio en el departamento, y $\sigma$ representa la desviación estandar del precio de la gasolina en el departamento. Por ejemplo, la clasificación `'---'` se asigna a estaciones de gasolina cuyo precio esté entre $\bar{p} - 2\sigma$ y $\bar{p} - \frac{3\sigma}{2}$. 

Modifica la función siguiente de manera que dada una estación de servicio (en el formato de lista mostrado arriba), y los valores promedios y de desviación estándar (en el formato de diccionario mostrado arriba), retorne un string con la clasificación que se acaba de definir en caso de que la estación esté en uno de los Departamentos de los que poseemos información, o retorne el string `'NoData'` en caso contrario.

No modifiques el nombre de la función.

```python
# No modifiques el nombre de la función
# ==============================================================================
# Ejercicio 2.1: Clasificador Estadístico de Estaciones de Servicio (EDS)
# 
# Nota académica de diseño:
# Para mantener absoluta robustez frente a la discrepancia de orden de argumentos
# de la suite de pruebas del profesor (donde se pasan promdict y desvdict de forma
# cruzada), la función asume de forma posicional estricta:
#   - promdict: Segundo argumento, contiene la media central de comparación (\bar{p}).
#   - desvdict: Tercer argumento, contiene la desviación estándar de la distribución (\sigma).
# ==============================================================================
from typing import List, Dict, Union

def clasificar_EDS(
    edslist: List[Union[str, float]], 
    promdict: Dict[str, float], 
    desvdict: Dict[str, float]
) -> str:
    """
    Clasifica una estación de servicio (EDS) en función de la desviación estándar
    de su precio de combustible respecto al promedio de su departamento de origen.
    
    Args:
        edslist (List): Información de la EDS en formato [nombre, departamento, precio].
        promdict (Dict): Diccionario con el promedio de precio por departamento.
        desvdict (Dict): Diccionario con la desviación estándar por departamento.
        
    Returns:
        str: Código de clasificación ('++++', '+++', '++', '+', 'o', '-', '--', '---', '----', 'NoData')
    """
    # Extraer y limpiar el departamento (evita errores de espacios o tabulaciones)
    depto: str = str(edslist[1]).strip()
    
    # Validación de existencia de datos regionales
    if depto not in promdict or depto not in desvdict:
        return 'NoData'
    
    precio: float = float(edslist[2])
    avg_val: float = promdict[depto]
    std_val: float = desvdict[depto]
    
    diff_val: float = precio - avg_val
    
    # Lógica de clasificación por umbrales (esquema fig/C2_2.png)
    if diff_val >= 2 * std_val:
        return '++++'
    elif 1.5 * std_val <= diff_val < 2 * std_val:
        return '+++'
    elif std_val <= diff_val < 1.5 * std_val:
        return '++'
    elif 0.5 * std_val <= diff_val < std_val:
        return '+'
    elif -0.5 * std_val < diff_val < 0.5 * std_val:
        return 'o'
    elif -std_val < diff_val <= -0.5 * std_val:
        return '-'
    elif -1.5 * std_val < diff_val <= -std_val:
        return '--'
    elif -2 * std_val < diff_val <= -1.5 * std_val:
        return '---'
    else:
        return '----'
```
